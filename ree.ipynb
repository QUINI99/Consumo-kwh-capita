## EVOLUCION DEL KWH/PER CAPITA

Datos obtenidos desde el Banco Mundial    https://data360.worldbank.org/en/search

```python
# prompt: crea df con kwh.csv

import pandas as pd

# Assuming kwh.csv is in the current directory. If not, provide the full path.
df = pd.read_csv('kwh.csv')

```

```python
# prompt: #selecciona registro con Country  Name=Alemania

# Select rows where 'Country Name' is 'Alemania'
china_data = df[df['Country Name'] == 'China']
alemania_data = df[df['Country Name'] == 'Alemania']
espana_data = df[df['Country Name'] == 'España']
india_data = df[df['Country Name'] == 'India']
japon_data = df[df['Country Name'] == 'Japón']
usa_data = df[df['Country Name'] == 'Estados Unidos']


```

```python
import matplotlib.pyplot as plt

# Transpose the data for easier plotting
china_data_transposed = china_data.set_index('Country Name').T
alemania_data_transposed = alemania_data.set_index('Country Name').T
espana_data_transposed = espana_data.set_index('Country Name').T
india_data_transposed = india_data.set_index('Country Name').T
japon_data_transposed = japon_data.set_index('Country Name').T
usa_data_transposed = usa_data.set_index('Country Name').T

# Select the columns from 1990 to 2024
years = [str(year) for year in range(1990, 2025)]
# Access the data using .loc to select rows by index labels
china_data_subset = china_data_transposed.loc[years]
alemania_data_subset = alemania_data_transposed.loc[years]
espana_data_subset = espana_data_transposed.loc[years]
india_data_subset = india_data_transposed.loc[years]
japon_data_subset = japon_data_transposed.loc[years]
usa_data_subset = usa_data_transposed.loc[years]

# Plot the data
plt.figure(figsize=(15, 6))
# Access the data using .iloc to select columns by position. Since 'España' is the only column, we use .iloc[:, 0]
plt.plot(china_data_subset.index, china_data_subset.iloc[:, 0],label="China")
plt.plot(alemania_data_subset.index, alemania_data_subset.iloc[:, 0], label="Alemania")
plt.plot(espana_data_subset.index, espana_data_subset.iloc[:, 0], label="España")
plt.plot(india_data_subset.index, india_data_subset.iloc[:, 0], label="India")
plt.plot(japon_data_subset.index, japon_data_subset.iloc[:, 0], label="Japón")
plt.plot(usa_data_subset.index, usa_data_subset.iloc[:, 0], label="Estados Unidos")
plt.xlabel("Año")
plt.ylabel("kWh")
plt.title("Consumo Kwh per cápita (1990-2024)")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

USA es el mayor consumidor
CHINA con un crecimiento de consumo muy elevado, causado por su alta industrialización.
ESPAÑA y el resto de economías europeas baja su consumo progresivamente. Europa se desindustrializa
